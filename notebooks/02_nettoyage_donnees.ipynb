```python
import pandas as pd 

df = pd.read_csv("../data/raw/customer_churn.csv")

df_clean = df.copy()

print("Nomvbre de clients:" , df_clean.shape[0])
```

```python
total_charges = df_clean["TotalCharges"].str.strip()

df_clean["TotalCharges"] = pd.to_numeric(
    total_charges,
    errors="coerce"
)

df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(0)

print("Type de la colonne :", df_clean["TotalCharges"].dtype)
print("Valeurs manquantes :", df_clean["TotalCharges"].isna().sum())
```

```python

print("Dimensions :", df_clean.shape)


print("\nValeurs manquantes :")
print(df_clean.isnull().sum())


print("\nDoublons :", df_clean.duplicated().sum())


df_clean.loc[
    df_clean["tenure"] == 0,
    ["customerID", "tenure", "TotalCharges", "Churn"]
]
```

```python
df_clean.to_csv(
    "../data/processed/costumer_churn_clean.csv",
    index=False
)

print("Dataset nettoyé enregistre avec succes ! ")
```
